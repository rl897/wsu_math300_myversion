```python
rng = np.random.default_rng(seed=0)
normal_sample_mean_5 = rng.normal(loc=5, scale=2, size=200)
#print(normal_sample_mean_5)
def count_above(arr, threshold):
    v=0
    for idx in arr:
        if (idx > threshold):
           v=v+1
    return v       
v=count_above(normal_sample_mean_5, 4)
print(v)

k=np.sum(normal_sample_mean_5 > 4)
print(k)
```
