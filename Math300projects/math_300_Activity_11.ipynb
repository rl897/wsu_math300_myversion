```python
import numpy as np
import matplotlib.pyplot as plt

print(plt.matplotlib.__version__)
```

```python
x=np.linspace(-10,10,500)
h=x*np.sin(x)

fig, ax =plt.subplots()
ax.plot(x,h, color="g", linewidth= .8)
ax.set_title("Plot of x*sin(x)")
ax.set_xlabel("x")
ax.set_ylabel("x*sin(x)")
ax.grid(True)
ax.axhline(0)
plt.show()
```

```python
k=np.linspace(0,4*np.pi, 500)

fig, axs = plt.subplots(nrows=2, ncols=3, figsize =(5,5),sharex=True)
axs[0,0].plot(k, np.sin(k), color="g")
axs[0,0].set_title("sin(k)")
axs[0,0].set_xlabel("k")
axs[0,0].set_ylabel("sin(k)")
axs[0,0].grid(True, alpha=0.3)

axs[0,1].plot(k, np.sin(2*k), color="g")
axs[0,1].set_title("sin(2k)")
axs[0,1].set_xlabel("k")
axs[0,1].set_ylabel("sin(2k)")
axs[0,1].grid(True, alpha=0.3)

axs[1,1].plot(k, np.sin(3*k), color="g")
axs[1,1].set_title("sin(3k)")
axs[1,1].set_xlabel("k")
axs[1,1].set_ylabel("sin(3k)")
axs[1,1].grid(True, alpha=0.3)

axs[1,0].plot(k, np.cos(k), color="g")
axs[1,0].set_title("cos(k)")
axs[1,0].set_xlabel("k")
axs[1,0].set_ylabel("cos(k)")
axs[1,0].grid(True, alpha=0.3)

axs[0,2].plot(k, np.cos(2*k), color="g")
axs[0,2].set_title("cos(2k)")
axs[0,2].set_xlabel("k")
axs[0,2].set_ylabel("cos(2k)")
axs[0,2].grid(True, alpha=0.3)

axs[1,2].plot(k, np.cos(3*k), color="g")
axs[1,2].set_title("cos(3k)")
axs[1,2].set_xlabel("k")
axs[1,2].set_ylabel("cos(3k)")
axs[1,2].grid(True, alpha=0.3)


plt.tight_layout()
plt.show()
```
